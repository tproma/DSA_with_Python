### Encapsulation

- restricting certain entities in class
- denoted by __ before the class variable
- call the variable using class name 
-- course2._iub_course__price

```python
class iub_course:
        def __init__(self, name, faculty, price):
                self.course_name = name
                self.faculty_name = faculty
                self.__price = price
        
        def display_course_details(self):
                print(f"Course Name: {self.course_name}")
                print(f"Faculty Name: {self.faculty_name}")
                print(f"Price: {self.price}")

        def update_price(self, new_price):
                self.price = new_price
                print(f"Price updated to: {self.price}")

        def enroll_student(self, student_name):
                print(f"{student_name} has been enrolled in {self.course_name}.")   
                

        def get_price(self):
                return self.__price
```

```python
course1 = iub_course("Python Programming I", "Proma", 5000)
course2 = iub_course("Web Development", "Md", 3000)
course3 = iub_course("Data Science", "Proma", 7000)
```

### Inheritance
- Inheriting everything from parent class

```python
class iub:
        
        def __init__(self, name, faculty, price):
                self.course_name = name
                self.faculty_name = faculty
                self.price = price

        def display_course(self):
                print(f"Course name: {self.course_name}")
                print(f"Faculty: {self.faculty_name}")
                print(f"Price: {self.price}")


class iub_course(iub):
        def __init__(self, name, faculty, price, duration):
                super().__init__(name, faculty, price)
                self.duration = duration
```

### Magic Method
- __init__():


```python
class iub:
    def __init__(self, name, faculty, price, student_count):
        self.name = name
        self.faculty = faculty
        self.price = price
        self.student_count = student_count
        

    def __str__(self):
        return f"Course Name: {self.name}, Faculty: {self.faculty}, Price: {self.price}"


    def __len__(self):
        return self.student_count

        
    def __repr__(self):
        return f"IUB({self.name}, {self.faculty},  {self.price},  {self.student_count})"
```

```python
obj = iub("DSA with Python", "Proma", 1000, 30)
print(obj)
print(len(obj))
print(repr(obj))
```

### Public , Private , Protected variable
- private : _name
- Protected : __name

```python
class iub:
    def __init__(self, name, faculty, price, student_count):
        self._name = name
        self._faculty = faculty
        self.price = price
        self._student_count = student_count
        

    @property # property Decorator
    def name(self):
        return self._name

    @property
    def faculty(self):
        return self._faculty

    @property
    def student_count(self):
        return self._student_count    


    @name.setter  # Setter Decorator
    def name(self, new_name):
        self._name = new_name  

    @name.deleter  # Deleter Decorator
    def name(self):
        del self._name
```

### Abstract method
- User will use it according to their will

```python
from abc import ABC, abstractmethod

class Iub(ABC):
    @abstractmethod
    def get_duration(self):
        pass

    @abstractmethod
    def get_price(self):
        pass    

class PythonCourse(Iub):
    def __init__(self):
        self.duration = "3 Months"
        self.price = 1000

    def get_duration(self):
        return f"{self.duration}"

    def get_price(self):
        return self.price   
    

class Data_Science_Course(Iub):
    def __init__(self):
        self.duration = "6 Months"
        self.price = 5000

    def get_duration(self):
        return f"Duration: {self.duration}"

    def get_price(self):
        return f"Price: ${self.price}"   
```

```python
obj = PythonCourse()
```

```python
obj.get_duration()
```

```python
obj.get_price()
```

```python
obj2 = Data_Science_Course()
```

```python
obj2.get_duration()
```
