### Encapsulation

```python
class iub_course:
        def __init__(self, name, faculty, price):
                self.course_name = name
                self.faculty_name = faculty
                self.price = price
        
        def display_course_details(self):
                print(f"Course Name: {self.course_name}")
                print(f"Faculty Name: {self.faculty_name}")
                print(f"Price: {self.price}")

        def update_price(self, new_price):
                self.price = new_price
                print(f"Price updated to: {self.price}")

        def enroll_student(self, student_name):
                print(f"{student_name} has been enrolled in {self.course_name}.")   
                
```

```python
course1 = iub_course("Python Programming I", "Proma", 5000)
course2 = iub_course("Web Development", "Md", 3000)
course3 = iub_course("Data Science", "Proma", 7000)
```
