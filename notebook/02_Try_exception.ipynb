```python
try:
    l = [2,3,4,5,6]
    l1 = []
    for i in l:
        l1.append(i**3)
    print(l1)
except Exception as e:
    print(e)
```

## Try Except Finally Block

```python
try:
    d = {"name": "Proma"}
    print(d["course"])
except Exception as e:
    print(e)

try:
    print("Tanjina Proma")
except Exception as e:
    print(e)
finally:
    print("This will always execute")
```

```python
try:
    d = {"name": "Proma"}
    print(d["course"])
except Exception as e:
    print(e)

try:
    print("Tanjina Proma")
except Exception as e:
    print(e)
finally:
    try:
        print("This will always execute")
    except Exception as e:
        print(e)
```

```python
try:
    a = int(input("a:"))
    b = int(input("b:"))
    c = a/b
    print(c)
    print(dsfkj)
except Exception as e:
    print(e)
```

```python
try:
    a = int(input("a:"))
    b = int(input("b:"))
    c = a/b
    print(c)
    print(dsflkghdlkfkj)
except ZeroDivisionError as e:
    print(e)
except NameError as e:
    print(e)
except Exception as e:
    print(e)
```

```python
try:
    print("Find even or Odd")
    n = int(input("n:"))
    try:
        if n%2 == 0:
            print("Even")
            try:
                print("if even then add 10")
                n +=10
                print(n)
            except Exception as e:
                print(e)
        else:
            print("Odd")
    
    except Exception as e:
        print(e)

except Exception as e:  
    print(e)
```
